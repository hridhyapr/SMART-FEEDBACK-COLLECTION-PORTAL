## Wheelchair Ramp Height Calculation

```python
import math

# Given values
ramp_length = 8  # meters
angle_degrees = 20 # degrees
angle_radians = math.radians(angle_degrees)

# Calculate the height using the sine function: height = length * sin(angle)
height = ramp_length * math.sin(angle_radians)

print(f"A wheelchair ramp of {ramp_length} meters long, making an angle of {angle_degrees} degrees with the ground, will reach a height of approximately {height:.2f} meters.")
```

## Drone Horizontal Distance Calculation

```python
import math

# Given values
cable_length = 50  # meters
angle_degrees = 45 # degrees

# Convert angle from degrees to radians for math.cos()
angle_radians = math.radians(angle_degrees)

# Calculate the horizontal distance using the cosine function: distance = length * cos(angle)
horizontal_distance = cable_length * math.cos(angle_radians)

print(f"For a drone with a {cable_length}m cable making a {angle_degrees} degree angle with the ground, the horizontal distance is approximately {horizontal_distance:.2f} meters.")
```

## Lighthouse Height Calculation

```python
import math

# Given values
distance_from_lighthouse = 80  # meters
angle_of_elevation_degrees = 35 # degrees

# Convert angle from degrees to radians for math.tan()
angle_of_elevation_radians = math.radians(angle_of_elevation_degrees)

# Calculate the height of the lighthouse using the tangent function: height = distance * tan(angle)
height_of_lighthouse = distance_from_lighthouse * math.tan(angle_of_elevation_radians)

print(f"For a boat 80m away from the lighthouse with an angle of elevation of {angle_of_elevation_degrees} degrees, the height of the lighthouse is approximately {height_of_lighthouse:.2f} meters.")
```

## Calculate Variance of Student Marks

## Calculate Median Salary

```python
import statistics
salaries = []
print("Enter employee salaries (type 'done' when finished):")

while True:
    salary_input = input("Enter salary: ")
    if salary_input.lower() == 'done':
        break
    try:
        salary = float(salary_input)
        salaries.append(salary)
    except ValueError:
        print("Invalid input. Please enter a number or 'done'.")

if not salaries:
    print("No salaries entered. Cannot calculate median.")
elif len(salaries) == 1:
    print(f"Only one salary entered: {salaries[0]}. The median is the salary itself.")
else:
    # Sort the list to find the median
    salaries.sort()
    n = len(salaries)

    if n % 2 == 1:  # Odd number of elements
        median_salary = salaries[n // 2]
    else:  # Even number of elements
        mid1 = salaries[n // 2 - 1]
        mid2 = salaries[n // 2]
        median_salary = (mid1 + mid2) / 2

    print(f"\nEmployee Salaries (sorted): {salaries}")
    print(f"The median salary is: {median_salary:.2f}")
```

## Most Frequently Purchased Product Category

```python
import statistics

category_ids = []
print("Enter product category IDs (type 'done' when finished):")

while True:
    category_input = input("Enter category ID: ")
    if category_input.lower() == 'done':
        break
    try:
        category_id = int(category_input)
        category_ids.append(category_id)
    except ValueError:
        print("Invalid input. Please enter an integer ID or 'done'.")

if not category_ids:
    print("No category IDs entered. Cannot determine the mode.")
else:
    try:
        most_frequent_category = statistics.mode(category_ids)
        print(f"\nAll Category IDs: {category_ids}")
        print(f"The most frequently purchased product category ID is: {most_frequent_category}")
    except statistics.StatisticsError:
        print(f"\nAll Category IDs: {category_ids}")
        print("No unique mode found. All categories appear with the same frequency or there are multiple modes.")
```

## Store Student Details in JSON File


```python
import json
import os

def register_student():
    """Accepts student details and saves them to students.json."""
    print("\n--- Enter New Student Details ---")
    student_id = input("Enter Student ID: ")
    student_name = input("Enter Student Name: ")
    student_course = input("Enter Course: ")
    student_email = input("Enter Email: ")
    student_phone = input("Enter Phone Number: ")

    new_student = {
        "ID": student_id,
        "Name": student_name,
        "Course": student_course,
        "Email": student_email,
        "Phone Number": student_phone
    }

    file_name = "students.json"
    all_students = []

    if os.path.exists(file_name) and os.path.getsize(file_name) > 0:
        try:
            with open(file_name, 'r') as file:
                all_students = json.load(file)
            if not isinstance(all_students, list):
                print(f"Warning: '{file_name}' does not contain a list. Initializing new list.")
                all_students = []
        except json.JSONDecodeError:
            print(f"Warning: '{file_name}' is corrupted or empty. Initializing new list.")
            all_students = []

    all_students.append(new_student)

    with open(file_name, 'w') as file:
        json.dump(all_students, file, indent=4)

    print(f"Student '{student_name}' successfully registered and saved to '{file_name}'.")

register_student()

print("\n--- Current content of students.json ---")
if os.path.exists("students.json") and os.path.getsize("students.json") > 0:
    with open("students.json", 'r') as file:
        print(file.read())
else:
    print("students.json is empty or does not exist yet.")
```

## Read Product Details from JSON File


```python
import json
import os

def display_product_info():
    """Reads products.json and displays product name, price, and stock."""
    file_name = "products.json"

    if not os.path.exists(file_name) or os.path.getsize(file_name) == 0:
        print(f"The file '{file_name}' does not exist or is empty. Please create it with product data first.")
        dummy_products = [
            {"name": "Laptop", "price": 1200.00, "stock": 15},
            {"name": "Mouse", "price": 25.50, "stock": 100},
            {"name": "Keyboard", "price": 75.00, "stock": 50}
        ]
        with open(file_name, 'w') as f:
            json.dump(dummy_products, f, indent=4)
        print(f"A dummy '{file_name}' has been created for demonstration purposes. Please run this cell again.")
        return

    try:
        with open(file_name, 'r') as file:
            products = json.load(file)

        if not isinstance(products, list):
            print(f"Error: The file '{file_name}' does not contain a list of products. Please check its format.")
            return

        print(f"\n--- Product Information from '{file_name}' ---")
        for product in products:
            name = product.get("name", "N/A")
            price = product.get("price", "N/A")
            stock = product.get("stock", "N/A")
            print(f"Product: {name}, Price: ${price:.2f}, Stock: {stock}")
        print("--------------------------------------------------")

    except json.JSONDecodeError:
        print(f"Error: Could not decode JSON from '{file_name}'. The file might be corrupted or malformed.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
display_product_info()
```

## Convert Dictionary to JSON with Sorted Keys
This program demonstrates how to take a Python dictionary and convert it into a JSON string where all the keys within objects are sorted alphabetically. This ensures consistency and readability, which is crucial for data interchange and configuration files.

```python
import json
employee_info = {
    "name": "Alice Wonderland",
    "employee_id": "EMP001",
    "department": "Engineering",
    "hire_date": "2023-01-15",
    "salary": 85000.00,
    "is_manager": False,
    "contact": {
        "email": "alice.w@example.com",
        "phone": "555-1234",
        "address": {
            "zip": "90210",
            "street": "123 Tech Lane",
            "city": "Innovation City"
        }
    }
}

print("\n--- Original Python Dictionary ---")
print(employee_info)
sorted_json_output = json.dumps(employee_info, indent=4, sort_keys=True)

print("\n--- JSON with Keys Sorted Alphabetically ---")
print(sorted_json_output)
```

## Store Multilingual Customer Information in JSON

```python
import json

customer_data = {
    "customer_id": "CUST007",
    "name_english": "John Doe",
    "name_hindi": "जॉन डो",
    "name_malayalam": "ജോൺ ഡോ",
    "name_tamil": "ஜான் டோ",
    "name_japanese": "ジョン・ドウ",
    "email": "john.doe@example.com",
    "address": {
        "street": "123 International St",
        "city": "Global City",
        "country": "Anywhere"
    },
    "preferences": {
        "language": "English",
        "notifications": True
    }
}

print("\n--- Original Python Dictionary with Multilingual Data ---")
print(customer_data)
multilingual_json_output = json.dumps(customer_data, indent=4, ensure_ascii=False)

print("\n--- JSON Output with Preserved Unicode Characters ---")
print(multilingual_json_output)
```
