```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
categorical_columns = [
    "season",
    "id",
    "city",
    "date"
]

for column in categorical_columns:
    print(f"\n--- {column.upper()} ---")
    print(df[column].value_counts(dropna=False))
```

```python
score_columns_raw = [
    "method",
    "result_margin",
    "player_of_match"
]

# Filter out columns that do not exist in the DataFrame
existing_score_columns = [col for col in score_columns_raw if col in df.columns]

print("--- Numeric Score Columns (Min/Max) ---")
for column in existing_score_columns:
    if pd.api.types.is_numeric_dtype(df[column]):
        print(
            f"{column}: "
            f"min={df[column].min()}, "
            f"max={df[column].max()}"
        )

print("\n--- Categorical Score Columns (Value Counts) ---")
for column in existing_score_columns:
    if pd.api.types.is_object_dtype(df[column]):
        print(f"\n--- {column.upper()} ---")
        # Use value_counts to inspect categorical data, including NaNs
        print(df[column].value_counts(dropna=False))
```

```python

clean_df = df.copy()

# Remove duplicate records
clean_df = clean_df.drop_duplicates()

# Convert numerical columns
for column in existing_score_columns:
    clean_df[column] = pd.to_numeric(
        clean_df[column],
        errors="coerce"
    )

print("Shape after duplicate removal:", clean_df.shape)
```

```python
print("Missing values before:")
print(clean_df[["result_margin", "target_runs", "target_overs"]].isnull().sum())

# Use median because it is less affected by extreme values.
clean_df["result_margin"] = clean_df["result_margin"].fillna(
    clean_df["result_margin"].median()
)

clean_df["target_runs"] = clean_df["target_runs"].fillna(
    clean_df["target_runs"].median()
)

clean_df["target_overs"] = clean_df["target_overs"].fillna(
    clean_df["target_overs"].median()
)

print("\nMissing values after:")
print(clean_df[["result_margin", "target_runs", "target_overs"]].isnull().sum())
```

```python
invalid_scores = {}

# Replace values in 'result_margin' not between 0 and 100 with 0
clean_df.loc[~clean_df['result_margin'].between(0, 100), 'result_margin'] = 0

for column in existing_score_columns:
    # Check if the column is numeric after the data type correction in previous steps
    if pd.api.types.is_numeric_dtype(clean_df[column]):
        # Recalculate invalid scores after cleaning
        invalid_scores[column] = (
            (~clean_df[column].between(0, 100)).sum()
        )
    else:
        # For non-numeric columns, consider 0 invalid scores for this check
        invalid_scores[column] = 0

display(
    pd.Series(invalid_scores, name="invalid_records")
)
```

```python
invalid_scores = {}

# The existing_score_columns from a previous cell was: ['method', 'result_margin', 'player_of_match']
# Now, we re-evaluate these based on their corrected state in clean_df.

# For 'result_margin': check if values are between 0 and 100. (already handled)
invalid_scores['result_margin'] = (~clean_df['result_margin'].between(0, 100)).sum()

# For 'method' and 'player_of_match': now they are object types with no NaN values,
# so for the purpose of the 'invalid score' numeric check, they will show 0 invalid records.
# For categorical columns, we consider 'invalid' as having NaN values or unexpected categories.
# Since we filled NaNs, there are no 'invalid' records from that perspective.

# To explicitly show 0 for non-numeric columns in this context:
invalid_scores['method'] = 0 # No NaNs, and not subject to numeric range check
invalid_scores['player_of_match'] = 0 # No NaNs, and not subject to numeric range check

display(pd.Series(invalid_scores, name='invalid_records'))

# Optionally, to confirm no NaNs in method and player_of_match:
print('\nConfirming no NaNs in method and player_of_match:')
print(f"Method NaNs: {clean_df['method'].isnull().sum()}")
print(f"Player_of_match NaNs: {clean_df['player_of_match'].isnull().sum()}")
```

```python
clean_df["performance_level"] = pd.cut(
    clean_df["result_margin"],
    bins=[0, 49, 59, 69, 79, 100],
    labels=[
        "Needs Significant Support",
        "Needs Improvement",
        "Average",
        "Good",
        "Excellent"
    ],
    include_lowest=True
)

display(clean_df.head())
```

```python
season_counts = clean_df["season"].value_counts()

plt.figure(figsize=(10, 5))
sns.barplot(
    x=season_counts.values,
    y=season_counts.index
)

plt.title("Matches by Season")
plt.xlabel("Number of Matches")
plt.ylabel("Season")
plt.show()
```
