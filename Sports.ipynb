```python
season = df.copy()
duplicates = season.sample(8, random_state=42)
df = pd.concat([season, duplicates], ignore_index=True)
```

```python
missing = (
    df.isnull()
      .sum()
      .to_frame("missing_count")
)

missing["missing_percent"] = (
    missing["missing_count"] / len(df) * 100
).round(2)

display(
    missing.sort_values(
        "missing_percent",
        ascending=False
    )
)
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
categorical_columns = [
    "season",
    "id",
    "city",
    "date"
]

for column in categorical_columns:
    print(f"\n--- {column.upper()} ---")
    print(df[column].value_counts(dropna=False))
```

```python
score_columns_raw = [
    "method",
    "result_margin",
    "player_of_match",
    "umpire1",
    "umpire2",
    "venue",
    "team1",
    "team2",
    "toss_winner",
    "toss_decision",
    "winner",
    "win_by_runs", # This column is not in df.info()
    "win_by_wickets", # This column is not in df.info()
    "player_of_match_score" # This column is not in df.info()
]

# Filter out columns that do not exist in the DataFrame
existing_score_columns = [col for col in score_columns_raw if col in df.columns]

print("--- Numeric Score Columns (Min/Max) ---")
for column in existing_score_columns:
    if pd.api.types.is_numeric_dtype(df[column]):
        print(
            f"{column}: "
            f"min={df[column].min()}, "
            f"max={df[column].max()}"
        )

print("\n--- Categorical Score Columns (Value Counts) ---")
for column in existing_score_columns:
    if pd.api.types.is_object_dtype(df[column]):
        print(f"\n--- {column.upper()} ---")
        # Use value_counts to inspect categorical data, including NaNs
        print(df[column].value_counts(dropna=False))
```
